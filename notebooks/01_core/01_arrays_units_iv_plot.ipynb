# Tutorial 01 — Arrays, units, and an I–V plot

This introductory notebook uses a **synthetic** current–voltage dataset to demonstrate a small but complete scientific workflow. The numerical values are illustrative and are not presented as experimental results.


## Learning objectives

By the end of the notebook, you should be able to:

- store measurements in NumPy arrays and a pandas table;
- convert milliamperes to amperes explicitly;
- estimate resistance from a linear fit;
- inspect residuals rather than reporting only a fitted value; and
- produce a labelled Matplotlib figure.


```python
import sys
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

print(f"Python: {sys.version.split()[0]}")
print(f"NumPy: {np.__version__}")
print(f"pandas: {pd.__version__}")
print(f"Matplotlib: {matplotlib.__version__}")
```

## 1. Define and inspect the data

The voltage is recorded in volts and the current in milliamperes. Keeping the original unit in the column name prevents an easy-to-miss factor-of-1000 error.


```python
voltage_V = np.array([0.50, 1.00, 1.50, 2.00, 2.50, 3.00])
current_mA = np.array([0.99, 2.02, 2.98, 4.03, 4.97, 6.01])

data = pd.DataFrame({
    "voltage_V": voltage_V,
    "current_mA": current_mA,
})
data
```

## 2. Convert units and fit a physical model

For an approximately ohmic device, \(V = RI + b\). The slope of a voltage-versus-current fit therefore has units of ohms. The intercept \(b\) can reveal a small offset.


```python
data["current_A"] = data["current_mA"] * 1e-3

slope_ohm, intercept_V = np.polyfit(
    data["current_A"], data["voltage_V"], deg=1
)

data["fitted_voltage_V"] = slope_ohm * data["current_A"] + intercept_V
data["residual_V"] = data["voltage_V"] - data["fitted_voltage_V"]

print(f"Estimated resistance: {slope_ohm:.2f} ohm")
print(f"Estimated intercept: {intercept_V:.4f} V")
print(f"Largest absolute residual: {data['residual_V'].abs().max():.4f} V")
data
```

## 3. Plot the measurements and fitted line


```python
fig, ax = plt.subplots(figsize=(6.4, 4.2))
ax.scatter(data["current_mA"], data["voltage_V"], label="Synthetic data", zorder=3)
ax.plot(data["current_mA"], data["fitted_voltage_V"], label="Linear fit")
ax.set_xlabel("Current (mA)")
ax.set_ylabel("Voltage (V)")
ax.set_title("Illustrative current–voltage analysis")
ax.grid(alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```

## Interpretation and checks

- The slope is the resistance because voltage was fitted as a function of current in amperes.
- A nearly constant slope and small residuals are consistent with the deliberately constructed ohmic example.
- A fit alone does not establish a physical mechanism. Experimental interpretation would also require device geometry, measurement conditions, uncertainty, and evidence about contact effects.
